```python
def price_to_pkr(price):
    price = str(price).lower().replace("pkr", "").strip()

    if price == "call for price":
        return None

    if "crore" in price:
        return float(price.replace("crore", "").strip()) * 10_000_000

    if "lacs" in price:
        return float(price.replace("lacs", "").strip()) * 100_000

    return None


price_values = df["Price"].apply(price_to_pkr)

price_values.describe()
```

```python
df.assign(
    price_pkr=price_values
)[
    ["nam", "Price", "Year", "Fuel", "Transmission", "Engine Capacity", "price_pkr"]
].sort_values(
    "price_pkr", ascending=False
).head(20)
```

```python
battery_kwh = pd.to_numeric(
    df.loc[
        df["Fuel"] == "Electric",
        "Engine Capacity"
    ].str.replace("kWh", "", case=False).str.strip(),
    errors="coerce"
)

battery_kwh.sort_values(ascending=False).head(20)
```

```python
ev_audit = df.loc[
    df["Fuel"] == "Electric",
    ["nam", "Engine Capacity", "Year", "Price"]
].copy()

ev_audit["battery_kwh"] = battery_kwh

ev_audit.loc[
    ev_audit["battery_kwh"] > 200
].sort_values(
    "battery_kwh", ascending=False
)
```

```python
pd.cut(
    battery_kwh,
    bins=[0, 50, 100, 150, 200, 500, 1000, float("inf")],
    right=True
).value_counts().sort_index()
```
